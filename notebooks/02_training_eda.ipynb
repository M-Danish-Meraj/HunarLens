**Strict Leakage Rule (§6.3, §7):** This notebook inspects **only** the 70% PSU-disjoint training partition (`data/splits/train.parquet`, `n = 17,913`). Validation and test partitions are never loaded.

```python
import json
from pathlib import Path
from IPython.display import Image, Markdown, display
import pandas as pd

REPO_ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
train_df = pd.read_parquet(REPO_ROOT / 'data/splits/train.parquet')
eda_summary = json.loads((REPO_ROOT / 'reports/metrics/eda_summary.json').read_text(encoding='utf-8'))
print(f"Training rows: {len(train_df):,} | Independent PSUs: {train_df['PCode'].nunique():,}")
```

```python
for plot_key in sorted(eda_summary['plots'].keys()):
    img_rel = eda_summary['plots'][plot_key]['plot_file']
    display(Image(filename=str(REPO_ROOT / img_rel)))

display(Markdown((REPO_ROOT / 'docs/eda_interpretations.md').read_text(encoding='utf-8')))
```
