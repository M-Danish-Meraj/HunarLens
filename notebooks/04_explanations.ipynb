```python
import json
import sys
from pathlib import Path
from IPython.display import Image, display
import pandas as pd

REPO_ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
if str(REPO_ROOT / 'src') not in sys.path:
    sys.path.insert(0, str(REPO_ROOT / 'src'))

from hunarlens.inference import HunarLensPredictor

imp = json.loads((REPO_ROOT / 'reports/metrics/global_feature_importance.json').read_text(encoding='utf-8'))
display(pd.DataFrame(imp['features']))
display(Image(filename=str(REPO_ROOT / 'reports/figures/08_global_permutation_importance.png')))
```

```python
predictor = HunarLensPredictor()
sample_profile = {
    'age': 22,
    'sex': 'Female',
    'marital_status': 'Never Married',
    'relationship_to_head': 'Son/Daughter',
    'can_read': 'Yes',
    'can_write': 'Yes',
    'highest_education': 'Degree & Postgraduate',
    'current_enrolment': 'Not Enrolled',
    'dependents_count': 3,
    'working_age_count': 3,
    'province_territory': 'Khyber Pakhtunkhwa',
    'locality': 'Urban',
}
res = predictor.predict_profile(sample_profile)
print('Estimated Probability:', res['estimated_probability'], '| Threshold:', res['decision_threshold'])
display(pd.DataFrame(res['local_explanation']['top_contributions']))
```
